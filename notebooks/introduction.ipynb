## What is ASE?

The **Atomic Simulation Environment (ASE)** is an open-source Python library for setting up, running, and analysing atomistic simulations {cite}`ase2017`. It was developed at the Technical University of Denmark (DTU) and is now maintained by a large community of contributors.

ASE is *not* a simulation engine itself — instead, it provides a common Python interface to many different calculators (electronic structure codes, force fields, machine-learned potentials) and a rich set of tools for building structures, running workflows, and analysing results.

```{figure} ../images/ase_overview.png
:name: ase-overview
:width: 70%
ASE acts as a flexible interface layer between your Python workflow and many different simulation backends.
```

### Why use ASE?

| Feature | Benefit |
|---------|---------|
| Consistent Python API | Same code works with GPAW, VASP, Quantum ESPRESSO, EMT, ... |
| Structure I/O | Read/write >50 file formats (CIF, POSCAR, XYZ, ...) |
| Visualisation | Built-in viewer; nglview integration for Jupyter |
| Workflow tools | Geometry optimisation, MD, NEB, transition state search |
| Open source | Freely available; easy to inspect and extend |

### ASE in the context of quantum optics research

In this course we apply ASE to materials relevant to **quantum and molecular optics** — an exciting area where solid-state physics meets quantum information science. Key material classes we will encounter include:

- **Diamond with NV centres** — optically addressable spin qubits and single-photon emitters
- **Hexagonal boron nitride (hBN)** — 2D material hosting ultra-bright single-photon emitters
- **III-nitride semiconductors (GaN, AlN)** — wide-bandgap hosts for quantum emitters and LEDs
- **Perovskites** — efficient light emitters and absorbers
- **Rare-earth doped materials** — telecom-wavelength quantum emitters


## ASE Structure

ASE is organised into several key modules:

```
ase/
├── Atoms          ← represents a molecule or crystal
├── io             ← reading and writing structure files
├── build          ← tools for constructing structures
├── calculators/   ← interfaces to simulation codes
├── optimize/      ← geometry optimisation algorithms
├── md/            ← molecular dynamics integrators
├── phonons        ← lattice dynamics
├── dft/           ← density functional theory utilities
└── visualize      ← structure visualisation
```

The central object in ASE is the **`Atoms`** class, which we will explore in detail in Lecture 2.

## The Calculator Concept

A **calculator** in ASE is an object that can compute physical properties (energy, forces, stress) for a given `Atoms` configuration. Calculators fall into three broad categories:

### 1. Built-in calculators
Pure Python implementations bundled with ASE. The **Effective Medium Theory (EMT)** potential is a fast interatomic potential for FCC metals — useful for testing workflows before running expensive DFT.

### 2. File-based calculators
ASE writes input files, calls an external code as a subprocess, and reads back the output. Supported codes include VASP, Quantum ESPRESSO, CASTEP, Abinit, and many others.

### 3. Machine-learned potentials
Modern neural-network and Gaussian approximation potentials (e.g. MACE, CHGNet, SevenNet) are increasingly used as fast surrogates for DFT. These are particularly exciting for large-scale simulations of defect environments relevant to quantum optics.


````python
# Import the EMT calculator (no external code needed)
from ase.calculators.emt import EMT
from ase.build import bulk

# Build an aluminium FCC structure
al = bulk('Al', 'fcc', a=4.05)
al.calc = EMT()

# Compute energy
energy = al.get_potential_energy()
print(f"Al total energy: {energy:.4f} eV")
print(f"Al energy per atom: {energy/len(al):.4f} eV/atom")
````
